## 전사 파일이란?
- 한 줄에 "**발화 하나**"가 들어 있는 텍스트 파일입니다.
- 형식: `발화ID  실제 말한 문장`  → 예) `104_003_0019 지난해 삼 월 ...`
- 같은 폴더에 같은 이름의 음성 파일(`104_003_0019.flac`)이 짝으로 들어 있습니다.

```python
# ===================== 여기만 바꾸면 됩니다 =====================
from pathlib import Path
TRANSCRIPT_PATH = Path(
    "/data/ASR/RAW/OpenSLR_SLR40_Zeroth_ko/test_data_01/003/112/112_003.trans.txt"
)
# ==============================================================

import re
import pandas as pd
from IPython.display import Audio, display

try:
    import soundfile as sf
except ImportError:
    sf = None
    print("ℹ soundfile 가 없어 '음성 길이/재생'은 건너뜁니다. 필요하면: pip install soundfile\n")

SEED      = 42     # 샘플을 무작위로 뽑되 매번 같은 결과가 나오게 고정
N_LISTEN  = 100    # 들어볼 최대 개수 (파일에 그보다 적으면 있는 만큼만)

AUDIO_DIR = TRANSCRIPT_PATH.parent   # 오디오(.flac)는 전사 파일과 '같은 폴더'에 있음

print("전사 파일 :", TRANSCRIPT_PATH)
print("오디오 폴더:", AUDIO_DIR)
if TRANSCRIPT_PATH.exists():
    print("→ 전사 파일을 찾았습니다 ✅")
else:
    print("→ ⚠ 파일이 없습니다. TRANSCRIPT_PATH 경로를 다시 확인하세요. (오타/권한/경로 변경)")
```

## 1. 전사 파일 읽어서 표로 만들기

파일을 한 줄씩 읽어 표(DataFrame)로 정리합니다.
발화ID `104_003_0019` 는 `화자_세션_순번` 구조라서, 여기서 **speaker / session / seq** 를 분리해 둡니다.
(이 표 `df` 를 아래 셀들이 계속 사용합니다.)

```python
# 전사 파일을 한 줄씩 읽어 표로 변환
rows = []
if TRANSCRIPT_PATH.exists():
    for line in TRANSCRIPT_PATH.read_text(encoding="utf-8", errors="replace").splitlines():
        parts = line.strip().split(maxsplit=1)   # 첫 공백 기준으로 'ID'와 '문장' 분리
        if len(parts) < 2:
            continue
        uid, text = parts[0], parts[1].strip()
        sp = uid.split("_")
        rows.append({
            "utterance_id": uid,
            "speaker": sp[0] if len(sp) > 0 else "",
            "session": sp[1] if len(sp) > 1 else "",
            "seq":     sp[2] if len(sp) > 2 else "",
            "text": text,
        })

df = pd.DataFrame(rows)

if len(df) == 0:
    print("⚠ 읽은 발화가 0개입니다. 파일이 비었거나 경로가 틀렸을 수 있어요.")
else:
    print(f"이 파일에는 발화(문장)가 총 {len(df)}개 있습니다.")
    print(f"화자(speaker): {sorted(df['speaker'].unique())}")
    print(f"세션(session): {sorted(df['session'].unique())}")
    print("\n[처음 3줄 미리보기]")
    for r in df.head(3).itertuples():
        print(f"  ID={r.utterance_id}  →  speaker={r.speaker}, session={r.session}, seq={r.seq}")
        print(f"     문장: {r.text}")
```

## 2. 숫자와 영어는 어떻게 적혀 있나?

- 숫자 예: `2018` → "이천 십 팔",  `5.16` → "오 쩜 일 육",  `70%` → "칠십 퍼센트"
- 영어 예: open → "오픈",  theme park → "테마파크"

 **아라비아 숫자(0-9)와 영어 알파벳(A-Z)이 실제로 들어있는지** 셉니다.

```python
if len(df):
    t = df["text"].fillna("")

    n_digit = int(t.str.contains(r"[0-9]").sum())
    n_latin = int(t.str.contains(r"[A-Za-z]").sum())
    print(f"아라비아 숫자(0-9)가 든 문장 : {n_digit} / {len(df)}")
    print(f"영어 알파벳(A-Z)이 든 문장   : {n_latin} / {len(df)}")
    if n_digit == 0 and n_latin == 0:
        print("→ 숫자·영어가 모두 '발음형 한글'로 적혀 있습니다.\n")
    else:
        print("→ 숫자/영어 원문이 일부 남아 있습니다. (정규화 규칙 확인 필요)\n")

    print("[발음형 숫자 표기 예시] (해당 단어가 든 문장을 하나씩)")
    for kw in ["쩜", "년", "월", "원", "퍼센트", "킬로미터"]:
        hit = df[t.str.contains(kw)]
        if len(hit):
            r = hit.iloc[0]
            print(f"  '{kw}' → {r['utterance_id']}: {r['text'][:55]}")
```

## 3. 특수 기호 분석 — ETRI 전사규칙(v1.0) 기준

ETRI 규칙이 정의한 기호가 이 전사 파일에 실제로 있는지 직접 셉니다.
규칙상 `/ ( ) * +` 는 **메타 기호로만** 쓰이므로, 일반 낭독 문장에는 없을 수 있습니다(0이면 정상).

- 음향정보: `b/`(숨) `l/`(웃음) `o/`(타인 말소리) `n/`(잡음)
- 전사규칙: 간투어 `아/ 그/ …`, `u/`(완전 불명확), `*`(끊김/불명확), `+`(반복)
- 표기형/발음형: 이중전사 `(철자)/(발음)`  ← 메타기호 `)/(` 존재 여부로 판단
- 참고: 숫자가 `(1999년)/(천 구백…)` 형태면 이중전사 유지, `천 구백…`만 있으면 발음형으로 정리된 것

```python
# 특수 기호 탐색: (A) ETRI 전사규칙 정의 기호  +  (B) 일반 괄호·기호 추가 점검
import re
if len(df):
    t = df["text"].fillna("")

    # (A) ETRI 전사규칙(v1.0)이 정의한 기호
    ETRI = {
        "b/  숨소리(breath)":          r"(?:(?<=\s)|^)b/",
        "l/  웃음(laugh)":             r"(?:(?<=\s)|^)l/",
        "o/  타인 말소리(문두)":        r"(?:(?<=\s)|^)o/",
        "n/  주변 잡음(noise)":         r"(?:(?<=\s)|^)n/",
        "u/  완전 불명확":              r"(?:(?<=\s)|^)u/",
        "간투어  아/ 그/ 어/ 음/ ...":   r"(?:(?<=\s)|^)(?:아|어|음|으|응|그|저|저기|에)/",
        "*   불명확/끊김/잡음동반":      r"\*",
        "+   반복·잘못 발성":           r"\+",
        ")/( 이중전사 (철자)/(발음)":    r"\)/\(",
        "문장부호  . ? ! ,":            r"[.?!,]",
    }

    # (B) 일반 괄호·기호 (ETRI 외 추가 점검) — 다른 데이터셋 호환용
    GENERIC = {
        "( ) 소괄호": r"[()]",
        "[ ] 대괄호": r"[\[\]]",
        "< > 꺾쇠":   r"[<>]",
        "{ } 중괄호": r"[{}]",
        "/  슬래시":  r"/",
        "@  골뱅이":  r"@",
        "#  샵":      r"#",
    }

    def scan(catalog, title):
        print(f"=== {title} ===")
        hit = {}
        for name, pat in catalog.items():
            c = int(t.str.count(pat).sum())
            hit[name] = c
            if c == 0:
                print(f"  {name:26s}: 현재 파일 기준 미발견")
            else:
                ex = df[t.str.contains(pat)]["text"].head(2).tolist()
                print(f"  {name:26s}: {c}회")
                for e in ex:
                    print(f"        예) {e[:60]}")
        return hit

    hit_etri    = scan(ETRI,    "(A) ETRI 전사규칙 정의 기호")
    print()
    hit_generic = scan(GENERIC, "(B) 일반 괄호·기호 추가 점검")

    # 메타 기호( / ( ) * + ) 총량 → 전사 표기가 살아있는지 한눈에 판단
    meta = int(t.str.count(r"[/()*+]").sum())
    print(f"\n메타 기호( / ( ) * + ) 총 {meta}회")
    if meta == 0:
        print("→ 메타 기호 0 = '정리된 단일 발음형' 전사 (이중전사/잡음태그 제거됨)")

    # 한글·공백이 아닌 글자 전체 (혹시 빠뜨린 기호 대비)
    others = t.str.findall(r"[^가-힣\s]").explode().value_counts()
    print("\n[한글·공백이 아닌 글자 전체]")
    print(others.head(20).to_string() if len(others) else "  없음 (한글 + 띄어쓰기로만 구성)")
```

## 4. 전사와 음성이 짝이 맞나? (매칭 검증)

전사 문장마다 같은 이름의 음성 파일이 있어야 합니다.
- 전사는 있는데 **음성이 없는** 경우
- 음성은 있는데 **전사가 없는** 경우
- 두 경우 모두 0이면 완벽한 1:1 매칭입니다.

```python
if len(df):
    # (1) 각 발화ID에 대응하는 .flac 이 같은 폴더에 있는지
    df["flac_path"]   = df["utterance_id"].map(lambda u: str(AUDIO_DIR / f"{u}.flac"))
    df["flac_exists"] = df["flac_path"].map(lambda p: Path(p).exists())
    missing_audio = df[~df["flac_exists"]]

    # (2) 폴더의 .flac 중 전사에 없는 것
    ids_in_trans = set(df["utterance_id"])
    flacs  = sorted(AUDIO_DIR.glob("*.flac"))
    orphan = [f.name for f in flacs if f.stem not in ids_in_trans]

    print(f"전사 문장 수    : {len(df)}")
    print(f"폴더의 .flac 수 : {len(flacs)}")
    print(f"\n전사 있음 / 음성 없음 : {len(missing_audio)}건")
    for r in missing_audio.head(5).itertuples():
        print(f"   - {r.utterance_id}")
    print(f"음성 있음 / 전사 없음 : {len(orphan)}건")
    for o in orphan[:5]:
        print(f"   - {o}")

    if len(missing_audio) == 0 and len(orphan) == 0:
        print("\n→ 전사와 음성이 1:1로 완벽히 맞습니다 ✅")
```

## 5. 실제 음성 들어보기

표시되는 플레이어의 **▶ 버튼**을 눌러 직접 듣고, 옆의 문장과 일치하는지 확인합니다.
- 자동 재생되지 않습니다 (원하는 것만 눌러 들으세요).
- `N_LISTEN`(Cell 2) 으로 개수를 조절할 수 있습니다. 파일에 든 발화가 그보다 적으면 있는 만큼만 나옵니다.

```python
if len(df) and sf is not None:
    tmp = df.copy()
    if "flac_path" not in tmp.columns:
        tmp["flac_path"] = tmp["utterance_id"].map(lambda u: str(AUDIO_DIR / f"{u}.flac"))
    tmp = tmp[tmp["flac_path"].map(lambda p: Path(p).exists())]

    n = min(N_LISTEN, len(tmp))
    if n == 0:
        print("⚠ 재생할 음성이 없습니다. (Cell 10에서 매칭 결과를 확인하세요)")
    else:
        sample = tmp.sample(n, random_state=SEED).sort_values("utterance_id")
        print(f"총 {len(tmp)}개 중 {n}개를 듣습니다 (seed={SEED}, 자동재생 아님)\n")
        for i, r in enumerate(sample.itertuples(), 1):
            try:
                dur = round(sf.info(r.flac_path).duration, 2)
            except Exception:
                dur = "?"
            print(f"[{i}/{n}] {r.utterance_id} · {dur}초")
            print(f"     문장: {r.text}")
            try:
                data, sr = sf.read(r.flac_path)
                display(Audio(data, rate=sr))   # autoplay 기본 꺼짐
            except Exception as e:
                print(f"     (재생 실패: {str(e)[:50]})")
            print()
elif sf is None:
    print("soundfile 미설치 → 재생 건너뜀. 'pip install soundfile' 후 다시 실행하세요.")
```
