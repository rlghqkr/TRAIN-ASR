# CallcenterQA 매핑 검증 — 전사 ↔ 음성 직접 대조

**가설(추정 매핑):** 라벨 `대화셋일련번호 SC6715` ↔ 오디오 `쇼핑_6715.m4a` (숫자부 일치)

이 노트북으로:
1. 라벨↔오디오 번호 교집합 확인 (매핑 후보 찾기)
2. 한 통화를 골라 **전사(문장 순서대로) 출력 + 음성 인라인 재생** → 들으면서 대조
   - 매핑이 맞는지 (말 내용이 전사와 같은 통화인지)
   - verbatim인지 (간투어·반복·맞장구까지 일치하는지, 아니면 다듬은 텍스트인지)

```python
from pathlib import Path
import json, re, zipfile, subprocess, tempfile, os
from IPython.display import Audio, display
try:
    import soundfile as sf
except ImportError:
    sf = None

BASE = Path("/data/ASR/RAW/AIHub_CallcenterQA_Audio/022.민원(콜센터)_질의-응답_데이터/01.데이터/1.Training")
LABEL_DIR = BASE / "라벨링데이터_231222_add"
AUDIO_DIR = BASE / "원천데이터_220125_add"
NUM = re.compile(r"(\d+)")

def fix_zipname(name):
    try:
        return name.encode("cp437").decode("cp949")
    except Exception:
        return name

print("LABEL_DIR:", LABEL_DIR.is_dir(), "| AUDIO_DIR:", AUDIO_DIR.is_dir())
```

## 1. 한 카테고리 로드 + 번호 교집합

> 기본: K쇼핑/반품 (라벨 해제돼 있음). 다른 카테고리는 CAT_LABEL/DOM/CAT만 바꾸면 됨.

```python
CAT_LABEL = LABEL_DIR / "민원(콜센터)_질의응답_K쇼핑_반품_Training/민원(콜센터) 질의응답_K쇼핑_반품_Training.json"
DOM, CAT  = "쇼핑", "반품"

# 라벨 로드 → 일련번호 숫자 집합
data = json.loads(CAT_LABEL.read_text(encoding="utf-8", errors="replace"))
label_nums = {}
for r in data:
    m = NUM.search(r.get("대화셋일련번호","") or "")
    if m: label_nums.setdefault(int(m.group(1)), r["대화셋일련번호"])
print(f"라벨: {len(data):,}행 / 고유 통화 {len(label_nums):,}개  예: {sorted(label_nums.values())[:5]}")

# 오디오 번호 수집 (해제 폴더 + zip 목록)
audio_loc = {}   # 번호 -> ("file", Path) 또는 ("zip", zip경로, 멤버명)
folder = AUDIO_DIR / DOM / CAT
if folder.is_dir():
    for ap in folder.iterdir():
        m = NUM.search(ap.stem)
        if m and ap.suffix.lower() in (".m4a",".mp3",".wav"):
            audio_loc[int(m.group(1))] = ("file", ap)
zp = AUDIO_DIR / DOM / f"{CAT}.zip"
if zp.exists():
    with zipfile.ZipFile(zp) as z:
        for raw in z.namelist():
            name = fix_zipname(raw)
            if not name.lower().endswith((".m4a",".mp3",".wav")): continue
            m = NUM.search(Path(name).stem)
            if m and int(m.group(1)) not in audio_loc:
                audio_loc[int(m.group(1))] = ("zip", zp, raw)
print(f"오디오: {len(audio_loc):,}개  번호범위 {min(audio_loc)}~{max(audio_loc)}" if audio_loc else "오디오 없음!")

inter = sorted(set(label_nums) & set(audio_loc))
print(f"\n교집합: {len(inter):,}개")
print("매핑 후보 (앞 10개):", inter[:10])
print("→ 아래 셀의 PICK에 이 중 하나를 넣고 실행")
```

## 2. 통화 1개: 전사 출력 + 음성 재생 (직접 대조)

> `PICK` 번호만 바꿔가며 여러 통화 확인. m4a는 ffmpeg로 임시 wav 변환 후 재생.

```python
PICK = 37   # 예: PICK = 6587  ← 위 교집합에서 하나 선택. None이면 첫 번째 자동

num = PICK if PICK is not None else (inter[0] if inter else None)
assert num is not None, "교집합이 비어 있음 — 매핑 가설이 틀렸거나 카테고리 불일치"

# --- 전사 재구성 (문장번호 순) ---
rows = [r for r in data if NUM.search(r.get("대화셋일련번호","") or "")
        and int(NUM.search(r["대화셋일련번호"]).group(1)) == num]
rows.sort(key=lambda r: int(r.get("문장번호","0") or 0))
print(f"=== 통화 {label_nums[num]} (오디오 번호 {num}) — {len(rows)}문장 ===\n")
for r in rows:
    text = (r.get("고객질문(요청)") or r.get("상담사질문(요청)") or
            r.get("고객답변") or r.get("상담사답변") or "").strip()
    print(f"[{r.get('문장번호','?'):>3}] {r.get('화자','?'):3s} | {text}")

# --- 오디오 가져와 재생 ---
loc = audio_loc[num]
if loc[0] == "file":
    src = loc[1]
else:
    tmpdir = Path(tempfile.mkdtemp(prefix="qa_check_"))
    with zipfile.ZipFile(loc[1]) as z:
        z.extract(loc[2], tmpdir)
    src = tmpdir / loc[2]
print(f"\n오디오: {src}")

# m4a/mp3 → wav(16k mono) 임시 변환 후 인라인 재생
wav = Path(tempfile.gettempdir()) / f"qa_check_{num}.wav"
subprocess.run(["ffmpeg","-y","-i",str(src),"-ar","16000","-ac","1",str(wav)],
               check=True, capture_output=True)
data_a, sr = sf.read(wav)
print(f"길이 {len(data_a)/sr:.1f}초 — 전사와 대조하며 들어보세요")
display(Audio(data_a, rate=sr))
```

```python
# === 원본 오디오 랜덤 청취 (셀 하나로 동작) ===
from pathlib import Path
from IPython.display import Audio, display
import random, subprocess, tempfile, soundfile as sf

BASE = Path("/data/ASR/RAW/AIHub_CallcenterQA_Audio/022.민원(콜센터)_질의-응답_데이터/01.데이터/1.Training")

# 해제돼 있는 오디오 전체에서 랜덤 (쇼핑 m4a + 다산콜센터 mp3 둘 다 포함)
pool = list(BASE.rglob("*.m4a")) + list(BASE.rglob("*.mp3"))
print(f"오디오 풀: {len(pool):,}개")

pick = random.choice(pool)          # 특정 파일을 듣고 싶으면: pick = Path("...경로...")
print("선택:", pick)

# ffmpeg로 임시 wav 변환 후 재생
wav = Path(tempfile.gettempdir()) / "listen_check.wav"
subprocess.run(["ffmpeg", "-y", "-i", str(pick), "-ar", "16000", "-ac", "1", str(wav)],
               check=True, capture_output=True)
data, sr = sf.read(wav)
print(f"길이: {len(data)/sr:.1f}초")
display(Audio(data, rate=sr))
```
