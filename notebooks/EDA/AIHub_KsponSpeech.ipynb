```python
from pathlib import Path
import re, numpy as np, pandas as pd
from collections import Counter
from IPython.display import Audio, display

# === KsponSpeech RAW 경로 ===
ROOT = Path("/data/ASR/RAW/AIHub_KsponSpeech/10.한국어음성")
SCRIPTS = ROOT / "KsponSpeech_scripts"

# PCM 규격 (KsponSpeech 표준: 16kHz, 16-bit, mono, headerless little-endian)
SR, SAMPW, CH = 16000, 2, 1

# KsponSpeech 전사/스크립트는 CP949 → cp949 우선 디코딩
def read_text(p, encs=("cp949", "utf-8", "euc-kr")):
    b = Path(p).read_bytes()
    for e in encs:
        try:
            return b.decode(e).strip()
        except Exception:
            continue
    return b.decode("utf-8", "replace").strip()

def pcm_dur(p):
    """헤더 없는 raw PCM → 파일 크기로 길이(초) 계산 (wav 헤더 읽기보다 훨씬 빠름)"""
    return Path(p).stat().st_size / (SR * SAMPW * CH)

def play_pcm(p, sr=SR):
    raw = Path(p).read_bytes()
    if len(raw) % 2:               # 홀수 바이트 PCM 보정 (자투리/잘림)
        raw = raw[:-1]
    pcm = np.frombuffer(raw, dtype=np.int16)
    return Audio(pcm, rate=sr)

print("ROOT 존재:", ROOT.exists())
for x in sorted(ROOT.iterdir()):
    print("  ", x.name, "(dir)" if x.is_dir() else "(file)")
```

## 1. 디렉토리 구조 / 파일 인벤토리

```python
# 1) train 오디오 폴더 한 개 들여다보기 (구조/이름 규칙 확인)
g = ROOT / "KsponSpeech_01"
subs = sorted([d for d in g.iterdir() if d.is_dir()])[:3]
print(f"{g.name} 하위 폴더 예시:", [d.name for d in subs])

leaf = subs[0]
print(f"\n{leaf.name} 안 파일 예시:")
for f in sorted(leaf.iterdir())[:6]:
    print("   ", f.name)

# 2) scripts 폴더 (.trn 전사 스크립트)
print("\nscripts 폴더:")
for f in sorted(SCRIPTS.iterdir()):
    print("   ", f.name, f"({f.stat().st_size/1024:.0f} KB)")
```

```python
# .trn 한 줄 형식 확인 (경로 :: 전사)
trn = SCRIPTS / "train.trn"
raw = read_text(trn).splitlines()
print(f"{trn.name}: {len(raw):,} 줄\n")
for line in raw[:3]:
    print(repr(line[:120]))

print("\n--- '::' 기준 파싱 ---")
for line in raw[:3]:
    path, text = line.split("::", 1)
    print("path:", path.strip())
    print("text:", text.strip(), "\n")
```

```python
# 모든 split의 .trn 로드 → 한 DataFrame (split, rel_path, text, abs_path)
SPLIT_FILES = {
    "train":      SCRIPTS / "train.trn",
    "dev":        SCRIPTS / "dev.trn",
    "eval_clean": SCRIPTS / "eval_clean.trn",
    "eval_other": SCRIPTS / "eval_other.trn",
}

rows = []
for split, fp in SPLIT_FILES.items():
    if not fp.exists():
        print(f"[경고] 없음: {fp.name}")
        continue
    for line in read_text(fp).splitlines():
        if "::" not in line:
            continue
        rel, text = line.split("::", 1)
        rows.append((split, rel.strip(), text.strip()))

df = pd.DataFrame(rows, columns=["split", "rel_path", "text"])
df["text"] = df["text"].astype("object")   # pyarrow string → object (Python re 사용)
df["abs_path"] = df["rel_path"].map(lambda r: str(ROOT / r))

print("split별 발화 수:")
print(df["split"].value_counts().to_string())
print(f"\n총 {len(df):,} 발화")
df.head()
```

```python
# 경로 구조 파싱: KsponSpeech_0X / KsponSpeech_XXXX / KsponSpeech_NNNNNN.pcm
pat = re.compile(r"(KsponSpeech_\w+)/(KsponSpeech_\w+)/([^/]+)$")
m = df["rel_path"].str.extract(pat)
df["group"], df["folder"], df["fname"] = m[0], m[1], m[2]
df["ext"] = df["fname"].str.extract(r"\.([^.]+)$")[0].str.lower()

print("파싱 성공률:", f"{df['fname'].notna().mean():.1%}")
print("\n확장자:", df["ext"].value_counts().to_dict())
print("\ngroup(최상위 폴더)별 분포:")
print(df["group"].value_counts().to_string())
print("\n파싱 실패 샘플(있다면):")
for p in df.loc[df["fname"].isna(), "rel_path"].head(5):
    print("   ", p)
```

```python
# 표본으로 ① 실제 pcm 존재 ② 옆 .txt가 .trn 전사와 일치하는지 점검
chk = df.sample(300, random_state=0)
miss_pcm = miss_txt = mismatch = 0
for r in chk.itertuples():
    pcm = Path(r.abs_path)
    if not pcm.exists():
        miss_pcm += 1; continue
    txt = pcm.with_suffix(".txt")
    if not txt.exists():
        miss_txt += 1; continue
    if read_text(txt) != r.text:
        mismatch += 1

print(f"표본 {len(chk)}건 점검")
print(f"  pcm 없음      : {miss_pcm}")
print(f"  옆 txt 없음   : {miss_txt}")
print(f"  trn != txt 불일치 : {mismatch}")
```

## 3. 전사

```python
txt = df["text"].fillna("")
N = len(txt)

# 백슬래시 들어가는 패턴은 f-string 밖에서 미리 계산
dual = txt.str.contains(r"\)/\(").mean() * 100      # (표기)/(발음)
eng  = txt.str.contains(r"[A-Za-z]").mean() * 100
han  = txt.str.contains(r"[一-龥]").mean() * 100

print(f"발화 수: {N:,}")
print(f"이중 전사 (표기)/(발음) 포함: {dual:.2f}%")
print(f"영문 포함: {eng:.2f}%  /  한자 포함: {han:.3f}%\n")

print("=== 슬래시 잡음 태그 / 마커 비율 (전체 기준) ===")
for tag, pat in [("b/", r"(?:^|\s)b/"), ("n/", r"(?:^|\s)n/"), ("l/", r"(?:^|\s)l/"),
                 ("o/", r"(?:^|\s)o/"), ("u/", r"(?:^|\s)u/"),
                 ("*", r"\*"), ("+", r"\+")]:
    hit = txt.str.contains(pat, regex=True)
    print(f"  {tag:>2} : {int(hit.sum()):>8,}건  ({hit.mean()*100:.2f}%)")
```

```python
# 1) 영문+슬래시 토큰 전수 (잡음 태그 vs 단위/약어 구분용)
print("=== 슬래시 토큰 전수 (상위 25) ===")
print(txt.str.findall(r"[A-Za-z]+/").explode().value_counts().head(25).to_string())

# 2) 한글/영문/숫자/공백 외 특수문자 전수 (다른 마커 탐지)
print("\n=== 특수문자 전수 (상위 30) ===")
print(txt.str.findall(r"[^가-힣a-zA-Z0-9\s]").explode().value_counts().head(30).to_string())
```

```python
# 자가수정/오발음 추정 마커(+, *)의 실제 모양 확인 + 오디오로 의미 검증
for sym in ["+", "*"]:
    pat = r".{0,15}" + re.escape(sym) + r".{0,15}"   # 마커 주변 문맥
    sub = txt[txt.str.contains(re.escape(sym))]
    print(f"=== '{sym}' 포함 {len(sub):,}건, 예시 10 ===")
    for t in sub.sample(min(10, len(sub)), random_state=0):
        mm = re.search(pat, t)
        print("  •", mm.group() if mm else t[:40])
    print()

print("샘플 오디오 3개 (전사와 함께):")
for r in df.sample(3, random_state=1).itertuples():
    print(f"[{r.split}] {r.text[:80]}")
    display(play_pcm(r.abs_path))
```

```python
# 확정 잡음 태그(b/ n/ l/ o/ u/)만 제거 후 영문 토큰 집계 (단위/약어는 유지)
clean = txt.str.replace(r"(?:^|\s)[bnlou]/", " ", regex=True)
has_eng = clean.str.contains(r"[A-Za-z]")
print(f"태그 제거 후 영문 포함: {int(has_eng.sum()):,}건 ({has_eng.mean()*100:.2f}%)\n")

ec = Counter()
for t in clean[has_eng].sample(min(50000, int(has_eng.sum())), random_state=1):
    ec.update(re.findall(r"[A-Za-z]+", t))
print("영문 토큰 상위 25:")
print(dict(ec.most_common(25)))

print("\n영문 포함 전사 샘플 10:")
for t in clean[has_eng].sample(min(10, int(has_eng.sum())), random_state=0):
    print("  •", t.strip()[:80])
```

## 3-1. 실제 음성 청취 — 라벨 검증

> 전사에 붙은 태그·기호가 **실제 오디오에서 무엇인지 귀로 확인**한다. 정규화 규칙(특히 `n/`의 의미, 이중전사 표기/발음 선택)을 데이터 보고 정하기 위한 근거.

```python
# 각 잡음 태그가 실제로 어떤 소리인지 귀로 확인 (b/숨 n/잡음 l/웃음 o/·u/는 미확인)
TAGS = {
    "b/": r"(?:^|\s)b/",
    "n/": r"(?:^|\s)n/",
    "l/": r"(?:^|\s)l/",
    "o/": r"(?:^|\s)o/",
    "u/": r"(?:^|\s)u/",
}
N_PER_TAG = 2   # 태그당 들어볼 개수

for tag, pat in TAGS.items():
    hits = df[df["text"].str.contains(pat, regex=True)]
    if hits.empty:
        print(f"[{tag}] 해당 없음\n"); continue
    print(f"=== '{tag}' 예시 {min(N_PER_TAG, len(hits))}개 (전체 {len(hits):,}건) ===")
    for r in hits.sample(min(N_PER_TAG, len(hits)), random_state=0).itertuples():
        print(f"  {r.text[:90]}")
        display(play_pcm(r.abs_path))
    print()
```

```python
# 이중 전사 (표기)/(발음) — 실제 발음이 어느 쪽인지 귀로 확인 → 정규화 mode 결정 근거
_DUAL = re.compile(r"\(([^)]*)\)/\(([^)]*)\)")
spell = lambda t: _DUAL.sub(lambda m: m.group(1), t)   # 표기형
pron  = lambda t: _DUAL.sub(lambda m: m.group(2), t)   # 발음형

dh = df[df["text"].str.contains(r"\)/\(", regex=True)]
print(f"이중전사 포함 {len(dh):,}건 중 4개 청취\n")
for r in dh.sample(min(4, len(dh)), random_state=2).itertuples():
    print(f"  원문  : {r.text[:90]}")
    print(f"  표기형: {spell(r.text)[:90]}")
    print(f"  발음형: {pron(r.text)[:90]}")
    display(play_pcm(r.abs_path))
    print()
```

```python
# 빈/아주 짧은 전사 — 오디오엔 말소리가 있는데 전사만 누락인지 확인 (라벨 갭 탐지)
cand = df[df["text_len"] <= 1]
print(f"전사 길이 <= 1 : {len(cand):,}건")
if cand.empty:
    cand = df.nsmallest(5, "text_len")   # 빈 게 없으면 최단 전사로 대체
    print("(빈 전사 없음 → 최단 전사로 대체)")
for r in cand.sample(min(5, len(cand)), random_state=0).itertuples():
    print(f"  [{r.text_len}자] {repr(r.text)}  {Path(r.abs_path).name}")
    display(play_pcm(r.abs_path))
```

## 4. 오디오 속성 (raw PCM)

> PCM은 헤더가 없어 **파일 크기 = 길이**. 209만 건 wav 헤더 읽기와 달리 stat만 하면 돼 매우 빠름.

```python
# split별 길이(초) 분포 — 대량 표본 (전수도 가능하나 표본이면 충분)
SAMPLE_N = 20000
dur_rows = []
for split, g in df.groupby("split"):
    s = g.sample(min(SAMPLE_N, len(g)), random_state=0)
    d = s["abs_path"].map(lambda p: pcm_dur(p) if Path(p).exists() else np.nan)
    dur_rows.append(pd.DataFrame({"split": split, "dur": d.values}))
dur = pd.concat(dur_rows, ignore_index=True).dropna()

g = dur.groupby("split")["dur"]
print("=== split별 길이(초) ===")
print(pd.DataFrame({
    "count": g.count(), "mean": g.mean(), "std": g.std(),
    "min": g.min(), "p50": g.median(), "p95": g.quantile(.95), "max": g.max(),
}).round(2).to_string())

print(f"\n표본 평균 길이: {dur['dur'].mean():.2f}s")
print(f"전체 총시간 추정: {dur['dur'].mean()*len(df)/3600:,.0f} 시간")
```

```python
# 가장 짧은/긴 발화 들어보기 (이상치·무음 점검)
samp = df.sample(min(30000, len(df)), random_state=2).copy()
samp = samp[samp["abs_path"].map(lambda p: Path(p).exists())]
samp["dur"] = samp["abs_path"].map(pcm_dur)

print("=== 가장 짧은 3개 ===")
for r in samp.nsmallest(3, "dur").itertuples():
    print(f"{r.dur:.2f}s | {r.text[:60]}")
    display(play_pcm(r.abs_path))

print("\n=== 가장 긴 3개 ===")
for r in samp.nlargest(3, "dur").itertuples():
    print(f"{r.dur:.2f}s | {r.text[:60]}")
    display(play_pcm(r.abs_path))
```

```python
# ============================================================
# 비식별화(PII) 토큰 포함 발화 → 전사 + 음성 직접 청취  (모든 EDA 노트북 공용)
# 파서로 DataFrame을 만든 셀을 먼저 실행한 뒤, 이 셀을 새 셀에 붙여 실행.
# DataFrame 변수(df/df_v/...)와 오디오 경로 컬럼(audio_path/wav/...)을 자동 탐지.
# ============================================================
import re
from pathlib import Path
from IPython.display import Audio, display
try:
    import soundfile as sf
except ImportError:
    sf = None
    print("⚠ soundfile 없음 → conda activate TRAIN-ASR")

import pandas as pd

# ---------- 설정 ----------
N_LISTEN = 5          # 들어볼 발화 수
RANDOM_STATE = 0      # None이면 매번 다른 표본
# 비식별화(PII) 토큰: 음향 토큰(b/ n/ l/ o/ u/)과 구분되는 익명화 전용 패턴
PII_PATTERNS = {
    "@ (이름 마커)":           re.compile(r"@"),
    "ㅇㅇ류 (익명화 2자+)":     re.compile(r"ㅇ{2,}"),
    "name/ (이름 태그)":        re.compile(r"(?:^|\s)name/", re.I),
    "[마스킹]":                re.compile(r"\[[^\]]{0,15}\]"),
    "<마스킹>":                re.compile(r"<[^>]{0,15}>"),
    "*** (별표 2+)":           re.compile(r"\*{2,}"),
    "xxx (엑스 2+)":           re.compile(r"[xX]{2,}"),
    "○○ (공백원 2+)":          re.compile(r"[○◯]{2,}"),
}

# ---------- 1) text DataFrame 자동 탐지 ----------
def _find_text_frame():
    g = globals()
    for name in ["df_v","df_valid","df","df_t","df_train","d","a"]:
        o = g.get(name)
        if isinstance(o, pd.DataFrame) and "text" in o.columns:
            return name, o
    for name, o in g.items():
        if not name.startswith("_") and isinstance(o, pd.DataFrame) and "text" in o.columns:
            return name, o
    return None, None

_name, _df = _find_text_frame()
if _df is None:
    raise RuntimeError("text 컬럼 DataFrame 없음 — 파서 셀을 먼저 실행하세요.")

# ---------- 2) 오디오 경로 컬럼 자동 탐지 ----------
AUDIO_COL = next((c for c in ["audio_path","wav","src_wav","audio","path","filepath"]
                  if c in _df.columns), None)
print(f"[대상] DataFrame '{_name}' · {len(_df):,} 발화 · 오디오 컬럼: {AUDIO_COL or '없음(PCM 직접 노트북일 수 있음)'}\n")

# ---------- 3) PII 토큰 집계 ----------
_txt = _df["text"].fillna("").astype("object")
print("=== 비식별화 토큰 집계 (text 원본) ===")
present = []
for label, pat in PII_PATTERNS.items():
    occ = int(_txt.str.count(pat).sum())
    utt = int(_txt.str.contains(pat).sum())
    if occ:
        present.append((label, pat, utt, occ))
        print(f"  {label:20s} 발화 {utt:>6,} · 출현 {occ:>6,} ({utt/len(_df)*100:.3f}%)")
if not present:
    print("  → 이 데이터셋엔 정의된 비식별화 토큰이 없음 (자유발화/낭독 등). 청취 생략.")

# ---------- 4) PII 포함 발화만 필터 → 전사 + 음성 재생 ----------
if present and AUDIO_COL:
    mask = pd.Series(False, index=_df.index)
    for _, pat, _, _ in present:
        mask |= _txt.str.contains(pat)
    hits = _df[mask]
    print(f"\n=== 비식별화 토큰 포함 발화 {len(hits):,}건 중 {min(N_LISTEN,len(hits))}개 청취 ===")
    print("   (전사의 마스킹 부분이 음성에서 실제로 어떻게 발화되는지 직접 확인)\n")
    sample = hits.sample(min(N_LISTEN, len(hits)), random_state=RANDOM_STATE)
    for r in sample.itertuples():
        text = getattr(r, "text", "")
        ap = getattr(r, AUDIO_COL, None)
        # 어떤 PII 패턴에 걸렸는지 표시
        tags = [lab for lab, pat, _, _ in present if pat.search(text or "")]
        print(f"[{', '.join(tags)}]")
        print(f"  전사: {text[:100]}")
        if sf and ap and Path(str(ap)).exists():
            try:
                data, sr = sf.read(str(ap))
                display(Audio(data, rate=sr))
            except Exception as e:
                print(f"   (재생 실패: {str(e)[:50]})")
        else:
            print(f"   (오디오 경로 없음/미존재: {ap})")
        print()
elif present and not AUDIO_COL:
    print("\n⚠ PII 토큰은 있으나 오디오 경로 컬럼을 못 찾음.")
    print("  이 노트북이 PCM을 직접 읽는 방식이면, 아래처럼 수동 지정:")
    print("  → sample = _df[mask].sample(N_LISTEN); 각 행의 키로 원본 PCM 경로를 구성해 재생")
```
