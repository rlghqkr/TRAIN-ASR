```python
playlist, missing_tsv, missing_wav = [], [], []
for k in TARGET_KEYS:
    row = index.get(k)
    if row is None:
        missing_tsv.append(k); continue
    wav = DATASET_ROOT / row["audio"]
    if not wav.exists():
        missing_wav.append(k); continue
    playlist.append((k, row, wav))

print(f"재생 가능 {len(playlist)} | transcript에 없음 {len(missing_tsv)} | wav 없음 {len(missing_wav)}")
if missing_tsv: print("  없는 key:", missing_tsv[:10])
if missing_wav: print("  wav 누락:", missing_wav[:10])

# ⚠️ 한 번에 다 띄우면 무거움 — playlist[:30] 식으로 끊어 권장
for i, (k, row, wav) in enumerate(playlist, 1):
    age  = row.get("age") or "-"
    gen  = row.get("gender") or "-"
    mask = "마스크" if row.get("has_mask") else ""
    star = "  ★star_fixed" if row.get("star_fixed") else ""
    print(f"[{i}/{len(playlist)}] {k}  |  {row['duration']}s  {age} {gen} {mask}{star}")
    print(f"  text      : {row['text']}")
    print(f"  text_norm : {row['text_norm']}")
    display(Audio(str(wav)))
    print("-" * 80)
```
